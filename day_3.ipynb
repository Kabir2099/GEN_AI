# call llm 

```python
from dotenv import load_dotenv
load_dotenv()
import os
```

```python
groq_api_key = os.getenv('groq_api_key')
```

```python
os.environ['groq_api_key'] = groq_api_key
```

```python
from langchain_groq import ChatGroq
llm_groq = ChatGroq(
    model = 'llama-3.1-8b-instant',
    temperature = 0.1,
    
)
```

```python
template = """
 summarize in the 2 lines given data : {data}
"""
```

```python
from langchain.prompts import PromptTemplate
```

```python
prompt = PromptTemplate(
    input_variables=['data'],
    template=template
)
```

```python
chain = prompt |llm_groq
```

```python
res = chain.invoke({'data': loaded_text})
```

```python
print(res.content)
```

# CSV Loader

```python
from langchain.document_loaders import CSVLoader
```

```python
import urllib.request
url = 'https://raw.githubusercontent.com/venkatareddykonasani/Datasets/master/Leads.csv'
file_name = 'Leads.csv'
urllib.request.urlretrieve(url, file_name)
```

```python
loaded = CSVLoader(file_name)
```

```python
loaded_csv = loaded.load()
```

```python
loaded_csv
```

```python
template = """ 
Read the following data and extract below information  and don't give me any code  just analyze the data :
- Maximum of the Promotion_Budget
- Maximum of the  Leads
-  Average of the Promotion_Budget
- Average of the Lead

Data - {input_data}

"""
```

```python
prompt = PromptTemplate(
    input_variables=['input_data'],
    template=template
)
```

```python
chain = prompt | llm_groq
```

```python
# have csv_agent
```

```python
res = chain.invoke({'input_data':loaded_csv})
```

```python
print(res.content)
```

# html Loader 

```python
from langchain.document_loaders import BSHTMLLoader
```

```python
import urllib.request
url ='https://raw.githubusercontent.com/venkatareddykonasani/Datasets/master/Word_bank_news_html/South_Asia_Global_Debt_Summary.html'
file_name = 'South_Asia_Global_Debt_Summary.html'
urllib.request.urlretrieve(url, file_name)
```

```python
loader = BSHTMLLoader(file_name, bs_kwargs={"features": "html.parser"})
```

```python
loaded_html_content = loader.load()
```

```python
print(loaded_html_content[0].page_content)
```
