# Text to Speech

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
groq_api_key = os.getenv('groq_api_key')
```

```python
from groq import Groq
client  = Groq(api_key=groq_api_key)

model = 'playai-tts'
text = 'I love Coding'
voice = "Fritz-PlayAI"
response_format = 'wav'
response = client.audio.speech.create(
    model=model,
    input=text,
    voice=voice,
    response_format=response_format)
```

```python
speech_file_path = 'speech_test.wav'
response.write_to_file(speech_file_path)
```

# Speech to Text ``

```python
filename = 'speech_test.wav'

with open(filename, 'rb') as f:
    transcript = client.audio.transcriptions.create(
        file=f,
        model = 'whisper-large-v3-turbo',
        prompt ='Transcribe the audio in the file.',
        response_format='verbose_json',
        timestamp_granularities = ["word", "segment"], # Optional (must set response_format to "json" to use and can specify "word", "segment" (default), or both)
        language="en",  # Optional
        temperature=0.0  # Optional
    )
```

```python
print(transcript.text)
```

# Image Vission 

```python

import base64
import os

# Function to encode the image
def encode_image(image_path):
  with open(image_path, "rb") as image_file:
    return base64.b64encode(image_file.read()).decode('utf-8')

# Path to your image
image_path = "Images\\Screenshot 2025-06-11 094426.png"

# Getting the base64 string
base64_image = encode_image(image_path)



chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "What's in this image?"},
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/jpeg;base64,{base64_image}",
                    },
                },
            ],
        }
    ],
    model="meta-llama/llama-4-scout-17b-16e-instruct",
)

print(chat_completion.choices[0].message.content)
```

# basic Chatbot 

```python
from langchain_groq import ChatGroq
llm_groq = ChatGroq(
    model = 'llama-3.1-8b-instant',
    temperature = 0.1,
    
)
```

```python
llm_groq.invoke("What is the capital of France?")
```

```python
user_input = input("Enter your question: ")

while user_input.lower() != 'exit':
    print(f"User input: {user_input}")
    response = llm_groq.invoke(user_input)
    print(f"Response: {response.content}")
    print("\n\n\n*************\n\n\n")
    user_input = input("Enter your question (or type 'exit' to quit): ")
```

# Sequntial chains

```python
template = '''
You are an elite literary curator with deep knowledge of global literature.

Your task is to generate a **concise list of 10 iconic and widely recognized books** that revolve around the central theme of **"{theme}"**.

⚠️ Important Instructions:
- Only list the **book titles** (no authors, no descriptions)
- Ensure the selections span **diverse genres, cultures, or time periods** if possible
- Keep the list **simple, clean, and numbered**

Output Format:
1. Book Title One  
2. Book Title Two  
...  
10. Book Title Ten
'''
```

```python
prompt = PromptTemplate(
    input_variables=["topic"],
    template=template
)
```

```python
print(prompt.format(theme="Horror"))
```

```python
book_chain = LLMChain(llm=llm_groq, prompt=prompt, output_key="book_list")
```

```python
res = book_chain.invoke({"theme": "Horror"})
```

```python
res
```

```python
book_summary_template = '''You are a master literary analyst.

From the provided list of books: ***"{book_list}"**, choose **one iconic title**.

Your task is to generate a **comprehensive and well-organized summary** of the selected book.

✍️ Instructions:
- Clearly **mention the selected book title**.
- Structure the summary into **three distinct sections**, each reflecting a major phase or theme of the book.
- Each section should contain **exactly three key summary points**.
- Use **clear, insightful language** suitable for an educated reader.

🎯 Goal: Provide a rich understanding of the book’s storyline, themes, and progression—without unnecessary filler.

Format Example:
**Selected Book**: Title of the Book

**Section 1: [Title/Theme of Section]**
1. Point one
2. Point two
3. Point three

**Section 2: ...**
...
'''
```

```python
book_summary_prompt = PromptTemplate(
    input_variables=["book_list"],
    template=book_summary_template
)
```

```python
print(book_summary_prompt.format(book_list="Python"))
```

```python
book_summray_chain = LLMChain(llm=llm_groq, prompt=book_summary_prompt, output_key="book_summary")
```

```python
res = book_summray_chain.invoke({"book_list": "Frankenstein"})
```

```python
print((res['book_summary']))
```

```python
from langchain.chains import SequentialChain
```

```python
book_chain = SequentialChain(chains=[book_chain, book_summray_chain],input_variables=["theme"], output_variables=["book_list", "book_summary"])
```

```python
res = book_chain.invoke({"theme": "Horror"})
```

```python
res
```

```python
print(res['book_list'])
```

```python
print(res['book_summary'])
```
