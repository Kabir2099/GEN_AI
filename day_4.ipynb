# Pydantatic Object 

````python
from pydantic import BaseModel
````

````python
class User(BaseModel):
    name : str
    age : int
````

````python
user = User(name ="Kabir", age= 12)
````

````python
print(user)
````

````python
from langchain.output_parsers import PydanticOutputParser
````

````python
from pydantic import Field
````

````python
class Scientist(BaseModel):
    name : str = Field("Name of the Scientist")
    bio : str = Field("4 lines summary about Scientist")
    dob : str = Field("Date of Birth")
    month : int = Field("Month of DOB in Integer")
````

````python
output_perser = PydanticOutputParser(pydantic_object= Scientist)
````

````python
print(output_perser.get_format_instructions())
````

````python
template = """ 
take the name of the scientist is {name} and try to fill below details-
{format_instruction}

**Instructions:**
- Output must strictly follow the JSON structure shown above.
- Do not include any additional text, commentary, or explanation outside the JSON.

"""
````

````python
prompt = PromptTemplate(
    input_variables=['name','format_instruction'],
    template= template
)
````

````python
chain = prompt | llm
````

````python
res = chain.invoke({"name":"Albert Einstein", "format_instruction": output_perser.get_format_instructions()})
````

````python
print(res.content)
````

````python
raw_output = res.content

if raw_output.startswith('```json'):
    raw_output = raw_output.removeprefix('```json').removesuffix('```').strip()
elif raw_output.startswith('```'):
    raw_output = raw_output.removeprefix('```').removesuffix('```')

output_json = json.loads(raw_output)
````

````python
output_json['month']
````
